## 1. NQ 데이터 다운로드 (~200MB)

retriever 학습용 JSON (train/dev) + 평가용 QA CSV (dev/test)

````python
import os
os.makedirs('/kaggle/working/data/nq', exist_ok=True)

NQ_BASE = 'https://dl.fbaipublicfiles.com/dpr/data/retriever'

# retriever 학습 데이터 (train/dev)
!wget -q -P /kaggle/working/data/nq {NQ_BASE}/biencoder-nq-train.json.gz
!wget -q -P /kaggle/working/data/nq {NQ_BASE}/biencoder-nq-dev.json.gz

# 압축 해제
!gunzip /kaggle/working/data/nq/biencoder-nq-train.json.gz
!gunzip /kaggle/working/data/nq/biencoder-nq-dev.json.gz

# 평가용 QA CSV
!wget -q -P /kaggle/working/data/nq {NQ_BASE}/nq-dev.qa.csv
!wget -q -P /kaggle/working/data/nq {NQ_BASE}/nq-test.qa.csv

!ls -lh /kaggle/working/data/nq
````
